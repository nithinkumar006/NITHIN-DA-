```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/car_price_performance.csv')
```

```python
df.head()
```

```python
brand_avg_price = df.groupby('Brand')['Price'].mean().sort_values()
brand_avg_price
```

```python
brand_mapping = {brand: i for i, brand in enumerate(brand_avg_price.index, start=1)}
brand_mapping
```

```python
encoded_brands = []
for b in df['Brand']:
    encoded_value = brand_mapping[b]
    encoded_brands.append(encoded_value)
```

```python
df['Encoded_Brand'] = encoded_brands
df[['Brand', 'Encoded_Brand']].drop_duplicates()
df
```

```python
min_brand = df['Encoded_Brand'].min()
max_brand = df['Encoded_Brand'].max()
df['Scaled_Brand'] = (df['Encoded_Brand'] - min_brand) / (max_brand - min_brand)
df
```

```python
min_hp = df['Horsepower'].min()
max_hp = df['Horsepower'].max()
df['Scaled_Horsepower'] = (df['Horsepower'] - min_hp) / (max_hp - min_hp)
df
```

```python
min = df['Fuel Efficiency'].min()
max = df['Fuel Efficiency'].max()
df['Scaled_Fuel_Efficiency'] = (df['Fuel Efficiency'] - min) / (max - min)
df
```

```python
df['Performance_Index'] = df['Horsepower'] * df['Fuel Efficiency']
df
```

```python
print("final_Table")
df
```
